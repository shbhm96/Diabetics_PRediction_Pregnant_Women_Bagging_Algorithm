```python
from sklearn import model_selection
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import numpy as np
import pandas as pd

datadf = pd.read_csv('/content/sample_data/diabetes.csv')
datadf.head()
```

```python
from sklearn.model_selection import train_test_split
array = datadf.values
X = array[:,0:8]
Y = array[:,8]
seed = 7
kfold = model_selection.KFold(n_splits=10, random_state=None)
cartClassifier = DecisionTreeClassifier()
x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=4)
cartClassifier.fit(x_train, y_train)
pred_value = cartClassifier.predict(x_test)
r2_score = cartClassifier.score(x_test, y_test)
print("Accuracy of Single Decision Tree model is",r2_score*100,"%")
```

```python
num_trees = 100
model = BaggingClassifier(estimator=cartClassifier, n_estimators=num_trees, random_state=seed)
results = model_selection.cross_val_score(model, X, Y, cv=kfold)
print("Accuracy of Each Bagging Decision Tree model is",results)
print("Accuracy of Bagging Decision Tree model is",results.mean()*100,"%")
```
